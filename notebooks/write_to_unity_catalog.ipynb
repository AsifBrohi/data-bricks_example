```python
"""
class that wraps the logic for writing to Unity Catalog
User inputs a dict:
    -volume location 
    -parquet file name 
    -method
    -table name 
    -catalog name 
    -schema name

"""
class UnityCatalogWriter:
    def __init__(self,params=None):
        if 'volume_path' in params:
            self.volume_path = params['volume_path']
        if 'parquet_file_name' in params:
            self.parquet_file_name = params['parquet_file_name']
        if 'method' in params:
            self.method = params['method']
        if 'table_name' in params:
            self.table_name = params['table_name']
        if 'catalog_name' in params:
            self.catalog_name = params['catalog_name']
        if 'schema_name' in params:
            self.schema_name = params['schema_name']
    

    def write_to_unity_catalog(self):
        try:
            if self.method=='overwrite':
                spark.read.parquet(f"{self.volume_path}/{self.parquet_file_name}.parquet").write.mode("overwrite").saveAsTable(f"{self.catalog_name}.{self.schema_name}.{self.table_name}")
                print(f"Table {self.table_name} overwritten")
            elif self.method=='append':
                spark.read.parquet(f"{self.volume_path}/{self.parquet_file_name}.parquet").write.mode("append").saveAsTable(f"{self.catalog_name}.{self.schema_name}.{self.table_name}")
                print(f"Table {self.table_name} appended")
            elif self.method=='upsert':
                spark.read.parquet(f"{self.volume_path}/{self.parquet_file_name}.parquet").write.mode("overwrite").option("mergeSchema", "true").saveAsTable(f"{self.catalog_name}.{self.schema_name}.{self.table_name}")
                print(f"Table {self.table_name} upserted")
            elif self.method=='update':
                spark.read.parquet(f"{self.volume_path}/{self.parquet_file_name}.parquet").write.mode("update").option("mergeSchema", "true").saveAsTable(f"{self.catalog_name}.{self.schema_name}.{self.table_name}")
                print(f"Table {self.table_name} updated")
            else:
                print("Invalid method")
        except Exception as e:
            print(e)
            print("Error writing to Unity Catalog")
            
```
